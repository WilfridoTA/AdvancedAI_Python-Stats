```python
# Canciones con mismo título pero artista diferente
duplicate_tracks_different_artists = df.groupby('Track').filter(lambda x: x['Artist'].nunique() > 1)
display(duplicate_tracks_different_artists.head())
```

```python
# Mismos artistas y títulos vs. lanzamientos distintos
duplicate_same_artist = df.groupby(['Track', 'Artist']).filter(lambda x: len(x) > 1)

# Priorización
def prioritize_release(group):
    # Nos quedamos con filas originales (no versiones de recopilación, deluxe, video, etc.)
    original_releases = group[~group['Album Name'].str.contains('Now|Deluxe|Single|Official Video', case=False, na=False)]

    # Se da prioridad al lanzamiento original
    if not original_releases.empty:
        return original_releases.sort_values(by='Release Date').iloc[0]
    else:
        return group.sort_values(by='Release Date').iloc[0]

# Aplicación de la priorización a los que comparten mismo título y artista
prioritized_duplicates = duplicate_same_artist.groupby(['Track', 'Artist']).apply(prioritize_release).reset_index(drop=True)

# Nuevo df sin duplicados y con las aplicaciones
duplicate_indices = duplicate_same_artist.index
non_duplicate_df = df.drop(duplicate_indices)

# Se combinan los df sin duplicados
cleaned_df = pd.concat([non_duplicate_df, prioritized_duplicates], ignore_index=True)
df = cleaned_df

display(df.head())
display(df.info())
```

```python
# Duplicados
remaining_duplicates = df[df.duplicated(subset=['Track', 'Artist', 'Album Name'], keep=False)].copy()

# Columnas con posibles 0s (numéricas)
numerical_cols = remaining_duplicates.select_dtypes(include=['float64', 'int64']).columns.tolist()

# 0s por fila
remaining_duplicates['zero_count'] = remaining_duplicates[numerical_cols].apply(lambda row: (row == 0).sum(), axis=1)

# Se orden por frecuencia de 0s y agrupan por duplicados (título - artista - álbum)
prioritized_by_zeros = remaining_duplicates.sort_values(by='zero_count').groupby(['Track', 'Artist', 'Album Name']).head(1)

# Df sin registros duplicados, pero sí priorizados
remaining_duplicate_indices = remaining_duplicates.index
non_remaining_duplicate_df = df.drop(remaining_duplicate_indices)

# Se combina el df limpio
cleaned_df_final = pd.concat([non_remaining_duplicate_df, prioritized_by_zeros.drop(columns='zero_count')], ignore_index=True)
df = cleaned_df_final

display(df.head())
display(df.info())
```

```python
# Título - Artista - Álbum
mask_album = df.duplicated(subset=['Track', 'Artist', 'Album Name'], keep=False)
remaining_duplicates_album = df.loc[mask_album]

# Título - Artista
mask_artist = df.duplicated(subset=['Track', 'Artist'], keep=False)
remaining_duplicates_artist = df.loc[mask_artist]

# Resultados por Título - Artista - Álbum
print("Duplicados por Título - Artista - Álbum:")
display(remaining_duplicates_album)

# Resultados por Título - Artista
print("\nDuplicados por Título - Artista:")
display(remaining_duplicates_artist)

print(f"\nNo. de Duplicados por Título - Artista - Álbum: {len(remaining_duplicates_album)}")
print(f"No. de Duplicados por Título - Artista: {len(remaining_duplicates_artist)}")
```

## Measures of Dispersion and Relative Variability

```python
# Desviación estándar
std_popularity = df["Spotify Popularity"].std()
std_streams = df["Spotify Streams"].std()
std_views = df["YouTube Views"].std()

# Media de popularity (para CV)
mean_popularity = df["Spotify Popularity"].mean()
mean_streams = df["Spotify Streams"].mean()
mean_views = df["Youtube Views"].mean()

# Coeficiente de variación (CV = std / mean)
cv_popularity = (std_popularity / mean_popularity) * 100
cv_streams = (std_streams / mean_streams) * 100
cv_views = (std_views / mean_views) * 100

print("\nStandard Deviation:")
print("Spotify Popularity:", std_popularity)
print("Spotify Streams:", std_streams)
print("YouTube Views:", std_views)

print("\nCoefficient of Variation (CV):")
print("Spotify Popularity: %", cv_popularity)
print("Spotify Streams: %", cv_streams)
print("YouTube Views: %", cv_views)
```

```python
plt.figure()
plt.bar(['Spotify Popularity', 'Spotify Streams', 'YouTube Views'], [cv_popularity, cv_streams, cv_views])
plt.title('Coeficientes de variación')
plt.ylabel('Coeficiente de Variación')
plt.show()
```

Como podemos ver en la gráfica de barras nuestros coeficientes tiene diferentes porcentajes, siendo la popularidad en Spotify la más pequeña con un CV de 52.73%, lo que nos indica que la probabilidad de que una canción sea popular es algo equitativa.

En Spotify streams podemos ver un brinco de un poco más del doble, lo que nos indica que existe una mayor variedad de posibles resultados para una canción, es decir, se vuelve muy complejo calcular si una canción tendrá muchos Streams o no, demostrando el dominio que tienen pocas canciones.

Por ultimo, podemos ver que el CV de Youtube es el más alto de todos. Relacionandolo con el comportamiento de las gráficas de arriba podemos concluir que la competencia de videos musicales es mucho mas agresiva/competitiva a diferencia de Spotify.


#Probabilistic Analysis with Bayes' Theorem

##  Defining Events

```python
df['Has_High_Popularity'] = (df['Spotify Popularity'] >= 85).astype(int)
df['Is_Explicit'] = (df['Explicit Track'] == 1).astype(int)
df
```

##Calculating Prior Probabilities

```python

p_A = df["Has_High_Popularity"].mean()
p_B = df["Explicit Track"].mean()
p_A_given_B = df.loc[df["Explicit Track"] == 1, "Has_High_Popularity"].mean()

print("P(A)  =", p_A)
print("P(B)  =", p_B)
print("P(A|B) =", p_A_given_B)
```

# Applying Bayes' Theorem

El **Teorema de Bayes** nos permite invertir probabilidades condicionales:

$$ P(B | A) = \frac{P(A | B) \cdot P(B)}{P(A)} $$

donde:

- **P(A):** Probabilidad de que una canción sea muy popular.  
- **P(B):** Probabilidad de que una canción sea explícita.  
- **P(A|B):** Probabilidad de que una canción sea muy popular dado que es explícita.  
- **P(B|A):** Probabilidad de que una canción sea explícita dado que es muy popular.  

Con el teorema de Bayes, pasamos de **P(A|B)** (ya conocida del dataset) a **P(B|A)**, usando la probabilidad total de A.


```python
# Teorema de Bayes P (B | A) = P(A | B) P(B) / P(A)
p_B_given_A = (p_A_given_B * p_B) / p_A

print(f"Si seleccionamos una canción al azar y resulta ser muy popular,\nsu probabilidad de que sea explícita es P(B|A) = {p_B_given_A:.4f}")

# Comparación
if p_B_given_A > p_B:
    print("\nSi una canción es muy popular, entonces es más probable que sea explícita.")
elif p_B_given_A < p_B:
    print("\nSi una canción es muy popular, entonces es menos probable que sea explícita.")
else:
    print("\nQue una canción sea muy popular NO cambia la probabilidad de que sea explícita.")
```
