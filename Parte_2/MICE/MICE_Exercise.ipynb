```python
import seaborn as sns
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

# **Exercise 1**

Use MICE on the life-expectancy dataset

```python
life_exp = pd.read_csv('Life Expectancy Data.csv')
life_exp
```

```python
print(life_exp.isnull().sum())
```

```python
life_exp_encoded = pd.get_dummies(life_exp, drop_first=True)
life_exp_encoded.head()
```

```python
imputer_le = IterativeImputer(max_iter=10, random_state=42)
```

```python
le_imputed = pd.DataFrame(imputer_le.fit_transform(life_exp_encoded), columns=life_exp_encoded.columns)
```

```python
print("Datos Originales")
print(life_exp_encoded.head())
print("Datos imputados")
print(le_imputed.head())
```

```python
planets = sns.load_dataset('planets')
planets.head()
```

```python
print(planets.isnull().sum())
```

```python
planets_encoded = pd.get_dummies(planets, drop_first=True)
planets_encoded.head()
```

```python
imputer_p = IterativeImputer(max_iter=10, random_state=42)
```

```python
planets_imputed = pd.DataFrame(imputer_p.fit_transform(planets_encoded), columns=planets_encoded.columns)
```

```python
print("Datos Originales")
print(planets_encoded.head())
print("Datos imputados")
print(planets_imputed.head())
```
